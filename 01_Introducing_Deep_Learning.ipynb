## 6. Pemanasan Python *(Tambahan)*

> Bagian 6–9 adalah **tambahan** (tidak ada kodenya di buku) untuk menyiapkan bekal Bab 3–6. Kode memakai contoh yang sama dengan bab-bab berikutnya.

Python yang dipakai di buku sangat sederhana: **list**, **loop**, dan **fungsi**.

```python
# List, loop, dan fungsi
toes = [8.5, 9.5, 9.9, 9.0]          # list angka

total = 0
for t in toes:                        # loop
    total += t
print('Total jari kaki:', total)

def rata_rata(daftar):                # fungsi
    return sum(daftar) / len(daftar)

print('Rata-rata:', rata_rata(toes))
```

**Fungsi `w_sum` (*weighted sum*)** akan menjadi inti prediksi neural network di Bab 3: mengalikan setiap input dengan bobotnya, lalu menjumlahkan hasilnya.

```python
def w_sum(a, b):
    assert(len(a) == len(b))          # panjang vektor harus sama
    output = 0
    for i in range(len(a)):
        output += (a[i] * b[i])
    return output

input_  = [8.5, 0.65, 1.2]            # toes, wlrec, nfans (pertandingan pertama)
weights = [0.1, 0.2, 0.0]
print(w_sum(input_, weights))
```

## 7. Pemanasan NumPy dan Benchmark Vektorisasi *(Tambahan)*

**NumPy** menyimpan angka dalam *array* dan menjalankan operasi pada seluruh array sekaligus (**vektorisasi**). Hasilnya sama dengan loop Python, tetapi jauh lebih cepat karena dikerjakan oleh kode C yang teroptimasi.

```python
import numpy as np

a = np.array([0, 1, 2, 3])           # vektor
c = np.array([[0, 1, 2, 3],
              [4, 5, 6, 7]])          # matriks

print(a * 0.1)                        # skalar x vektor
print(c * 0.2)                        # skalar x matriks
print(np.array([8.5, 0.65, 1.2]).dot(np.array([0.1, 0.2, 0.0])))   # dot product = w_sum
```

```python
import time

x = np.random.rand(1_000_000)
w = np.random.rand(1_000_000)

t0 = time.perf_counter()
hasil_loop = 0.0
for i in range(len(x)):
    hasil_loop += x[i] * w[i]
t_loop = time.perf_counter() - t0

t0 = time.perf_counter()
hasil_np = x.dot(w)
t_np = time.perf_counter() - t0

print(f'Loop Python : {hasil_loop:.4f}  ({t_loop:.4f} detik)')
print(f'NumPy dot   : {hasil_np:.4f}  ({t_np:.6f} detik)')
print(f'Hasil sama  : {np.isclose(hasil_loop, hasil_np)} | NumPy ~{t_loop / t_np:,.0f}x lebih cepat')
```

**Interpretasi:** kedua cara menghasilkan nilai yang sama, tetapi vektorisasi NumPy berkali-kali lipat lebih cepat. Itulah alasan semua neural network di buku ini memakai operasi matriks.

## 8. Matematika SMA yang Dibutuhkan *(Tambahan)*

Tiga konsep cukup: **perkalian**, **grafik x–y**, dan **slope (kemiringan)**.

- Perkalian: `prediksi = input × bobot`.
- Grafik x–y: menggambarkan hubungan dua kuantitas (mis. bobot vs error).
- **Slope** garis $= \dfrac{\Delta y}{\Delta x}$, yaitu seberapa cepat $y$ berubah bila $x$ berubah. Konsep ini menjadi dasar ***derivative*** pada Bab 4 (slope kurva error).

```python
import matplotlib.pyplot as plt

x = np.linspace(0, 4, 50)
y = 0.5 * x + 1                       # garis dengan slope 0.5

plt.figure(figsize=(5, 3.5))
plt.plot(x, y)
plt.plot([1, 3], [1.5, 1.5], 'r--'); plt.plot([3, 3], [1.5, 2.5], 'r--')
plt.text(2, 1.25, 'Δx = 2'); plt.text(3.05, 2.0, 'Δy = 1')
plt.title('Slope = Δy / Δx = 0.5')
plt.xlabel('x'); plt.ylabel('y'); plt.grid(alpha=0.3)
plt.show()
```

## 9. Sekilas: Neural Network yang *Belajar* *(Tambahan)*

Sebagai gambaran "hadiah" di Bab 4, berikut neural network **satu bobot** yang belajar sendiri: ia menebak, membandingkan dengan target, lalu menggeser bobotnya. Penjelasan lengkap ada di Bab 3 dan 4.

```python
weight = 0.5
input_ = 0.5
goal   = 0.8

for i in range(12):
    pred  = input_ * weight
    error = (pred - goal) ** 2
    weight -= (pred - goal) * input_      # geser bobot ke arah yang mengurangi error
    print(f'iterasi {i+1:2d} | prediksi = {pred:.4f} | error = {error:.6f}')
```

**Interpretasi:** error terus mengecil dan prediksi mendekati target 0,8. Ini intisarinya: **jaringan belajar dengan mengurangi error**.
